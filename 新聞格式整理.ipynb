```python
import pandas as pd
import ast
import time
```

```python
df1 = pd.read_excel("新聞資料/經濟日報0520.xlsx")
df2 = pd.read_excel("新聞資料/鉅亨網0520.xlsx")
df3 = pd.read_excel("新聞資料/moneyDJ0520.xlsx")
```

### 經濟日報

```python
# ===處理===
# Function to convert text to a list, removing empty lines
def text_to_list(text):
    return [item for item in text.split('\n') if item.strip()]
def textlist_to_list(text):
    return ast.literal_eval(text)

def adjust_change_line(text):
    return text.replace('\n\n\n\n\n\n\n\n\n', '').replace('\n\r\n', '\n')

##
df1.drop(columns='link', inplace= True)
df1 = df1.dropna(subset=['content'])
# Different format
# df1['keywords'] = df1['keywords'].apply(lambda x: text_to_list(x) if pd.isna(x) == False else x)
df1['keywords'] = df1['keywords'].apply(lambda x: textlist_to_list(x) if pd.isna(x) == False else x)
##
df1['content'] = df1['content'].apply(lambda x: adjust_change_line(x) if pd.isna(x) == False else x)
##
df1['content'] = df1['content'].apply(lambda x: x[1:] if x[0] == '\n' else x)
```

### 鉅亨網

```python
df0501 = pd.read_excel("新聞資料/鉅亨網0501.xlsx")
df0512 = pd.read_excel("新聞資料/鉅亨網0512.xlsx")
df0519 = pd.read_excel("新聞資料/鉅亨網0519.xlsx")
df0520 = pd.read_excel("新聞資料/鉅亨網0520.xlsx")
df0521 = pd.read_excel("新聞資料/鉅亨網0521.xlsx")
df0523 = pd.read_excel("新聞資料/鉅亨網0523.xlsx")
df0524 = pd.read_excel("新聞資料/鉅亨網0524.xlsx")
df0525 = pd.read_excel("新聞資料/鉅亨網0525.xlsx")
df0526 = pd.read_excel("新聞資料/鉅亨網0526.xlsx")
df0528 = pd.read_excel("新聞資料/鉅亨網0528.xlsx")
df0530 = pd.read_excel("新聞資料/鉅亨網0530.xlsx")
df0531 = pd.read_excel("新聞資料/鉅亨網0531.xlsx")
df0603 = pd.read_excel("新聞資料/鉅亨網0603.xlsx")
df_all = pd.concat([df0501, df0512, df0519, df0520, df0521, df0523, df0524, df0525, df0526, df0528, df0530, df0531, df0603], axis = 0)
df_all.reset_index(drop=True, inplace=True)
df_all.head()
```

```python
df_all.columns
```

```python
def delete_redundant_text(text):
    keyword = 'figure'
    text = text.split(keyword)[0]
    return text.replace('&lt;', '').replace('strong&gt;', '').replace('hr&gt;', '').replace('h6&gt;', '').replace('/p&gt;', '').replace('p&gt;', '').replace('&nbsp;', '').replace('br&gt;', '').replace('em&gt;', '').replace('//', '').replace("''", '')

df_all['time'] = df_all['publishAt'].apply(lambda x: time.strftime("%Y-%m-%d %H:%M:%S", time.localtime(x)))
                                     
df_all.drop(columns=['newsId', 'hasCoverPhoto', 'isIndex', 'isCategoryHeadline', 'video', 'payment', 'feature',
       'otherProduct', 'source', 'isOutsource', 'keyword', 'is24h',
       'publishAt', 'coverSrc', 'abTesting', 'categoryId', 'categoryName',
       'columnists', 'fundCategoryAbbr', 'fbShare', 'fbComment',
       'fbCommentPluginCount'], inplace= True)
df_all['news'] = ['鉅亨網']*df_all.shape[0]
df_all['content'] = df_all['content'].apply(lambda x: delete_redundant_text(x))
df_all.dropna(subset=['title', 'content', 'time'], inplace=True)
# Desired order of columns
new_order = ['news', 'time', 'title', 'summary', 'content', 'stock', 'etf', 'market']
# Reorder the columns
df_all = df_all[new_order]
```

```python
df_all = df_all.tail(100).reset_index(drop = True)
```

```python
df_all
```

```python
df_all.to_excel('整理過後的新聞資料/鉅亨網100.xlsx', index=False)
```

```python
# ===處理===
def adjust_change_line(text):
    return text.replace('\n\n\n\r\n', '\n')


df3.rename(columns={'article': 'content'}, inplace=True)
df3['drop'] = df3['content'].apply(lambda x: 0 if 'MoneyDJ新聞' in x else 1)
df3['title'] = df3['title'].apply(lambda x: x.replace('\n', ''))
df3['content'] = df3['content'].apply(lambda x: adjust_change_line(x) if pd.isna(x) == False else x)
df3 = df3[df3['drop'] == 0].reset_index(drop=True)
df3.drop(columns=['link', 'drop'], inplace= True)
df3['content'] = df3['content'].apply(lambda x: x[39:])
df3['content'] = df3['content'].apply(lambda x: x[1:] if x[0] == '\n' else x)
```
